```python

import numpy as np
import matplotlib.pyplot as plt

# =====================================
# 1. DEFINIR OS COEFICIENTES
# =====================================

a = 1
b = -5
c = 6

# Verificar se é uma equação do segundo grau
if a == 0:
    raise ValueError("O coeficiente 'a' não pode ser zero.")

# =====================================
# 2. DEFINIR A FUNÇÃO
# =====================================

def f(x):
    return a*x**2 + b*x + c

# =====================================
# 3. CALCULAR O DELTA E AS RAÍZES
# =====================================

delta = b**2 - 4*a*c

if delta > 0:
    x1 = (-b + np.sqrt(delta)) / (2*a)
    x2 = (-b - np.sqrt(delta)) / (2*a)
    print("Duas raízes reais:")
    print(f"x1 = {x1:.4f}")
    print(f"x2 = {x2:.4f}")

elif delta == 0:
    x1 = -b / (2*a)
    print("Uma raiz real:")
    print(f"x = {x1:.4f}")

else:
    print("A equação não possui raízes reais.")

# =====================================
# 4. CALCULAR O VÉRTICE
# =====================================

xv = -b / (2*a)
yv = f(xv)

print(f"\nDelta = {delta}")
print(f"Vértice: ({xv:.4f}, {yv:.4f})")

# =====================================
# 5. GERAR OS DADOS DO GRÁFICO
# =====================================

x = np.linspace(xv - 10, xv + 10, 500)
y = f(x)

# =====================================
# 6. CONSTRUIR O GRÁFICO
# =====================================

plt.figure(figsize=(10, 6))

plt.plot(x, y, label=f'f(x) = {a}x² + ({b})x + ({c})',
         linewidth=2)

# Marcar o vértice
plt.scatter(xv, yv, color='red', s=80,
            label=f'Vértice ({xv:.2f}, {yv:.2f})',
            zorder=5)

# Marcar as raízes reais
if delta > 0:
    plt.scatter([x1, x2], [0, 0], color='green',
                s=70, label='Raízes', zorder=5)

elif delta == 0:
    plt.scatter(x1, 0, color='green',
                s=70, label='Raiz', zorder=5)

# Eixos cartesianos
plt.axhline(0, color='black', linewidth=0.8)
plt.axvline(0, color='black', linewidth=0.8)

# Configurações visuais
plt.title('Gráfico da Função do Segundo Grau',
          fontsize=15)
plt.xlabel('Eixo X', fontsize=12)
plt.ylabel('Eixo Y', fontsize=12)
plt.grid(True, linestyle='--', alpha=0.5)
plt.legend()
plt.tight_layout()

# Exibir gráfico
plt.show()
```

```python

import numpy as np
import matplotlib.pyplot as plt

# =====================================
# 1. DEFINIR OS COEFICIENTES
# =====================================

a = 1
b = -6
c = 11
d = -6

# Verificar se é uma equação do terceiro grau
if a == 0:
    raise ValueError("O coeficiente 'a' não pode ser zero.")

# =====================================
# 2. DEFINIR A FUNÇÃO
# =====================================

def f(x):
    return a*x**3 + b*x**2 + c*x + d

# =====================================
# 3. CALCULAR AS RAÍZES
# =====================================

coeficientes = [a, b, c, d]
raizes = np.roots(coeficientes)

print("RAÍZES DA EQUAÇÃO:")
for i, raiz in enumerate(raizes, 1):
    print(f"x{i} = {raiz:.4f}")

# Identificar raízes reais
raizes_reais = [
    raiz.real for raiz in raizes
    if abs(raiz.imag) < 1e-9
]

print("\nRaízes reais:")
for raiz in raizes_reais:
    print(f"x = {raiz:.4f}")

# =====================================
# 4. CALCULAR OS PONTOS CRÍTICOS
# =====================================

# Derivada: f'(x) = 3ax² + 2bx + c
pontos_criticos = np.roots([3*a, 2*b, c])

pontos_reais = [
    ponto.real for ponto in pontos_criticos
    if abs(ponto.imag) < 1e-9
]

print("\nPONTOS CRÍTICOS:")
for ponto in pontos_reais:
    print(f"x = {ponto:.4f}, y = {f(ponto):.4f}")

# =====================================
# 5. CALCULAR O PONTO DE INFLEXÃO
# =====================================

xi = -b / (3*a)
yi = f(xi)

print("\nPONTO DE INFLEXÃO:")
print(f"Xi = {xi:.4f}")
print(f"Yi = {yi:.4f}")

# =====================================
# 6. GERAR OS DADOS DO GRÁFICO
# =====================================

x_min = min(-5, min(raizes_reais) - 2)
x_max = max(5, max(raizes_reais) + 2)

x = np.linspace(x_min, x_max, 1000)
y = f(x)

# =====================================
# 7. CONSTRUIR O GRÁFICO
# =====================================

plt.figure(figsize=(11, 7))

plt.plot(
    x, y,
    label=f'f(x) = {a}x³ + ({b})x² + ({c})x + ({d})',
    linewidth=2
)

# Eixos cartesianos
plt.axhline(0, color='black', linewidth=0.8)
plt.axvline(0, color='black', linewidth=0.8)

# Marcar raízes reais
for i, raiz in enumerate(raizes_reais):
    plt.scatter(
        raiz, 0,
        color='green',
        s=70,
        label='Raízes reais' if i == 0 else None,
        zorder=5
    )

# Marcar pontos críticos
for i, ponto in enumerate(pontos_reais):
    plt.scatter(
        ponto, f(ponto),
        color='orange',
        s=80,
        label='Pontos críticos' if i == 0 else None,
        zorder=5
    )

# Marcar ponto de inflexão
plt.scatter(
    xi, yi,
    color='red',
    s=80,
    label='Ponto de inflexão',
    zorder=5
)

# Configurações visuais
plt.title('Gráfico da Função do Terceiro Grau', fontsize=15)
plt.xlabel('Eixo X', fontsize=12)
plt.ylabel('Eixo Y', fontsize=12)
plt.grid(True, linestyle='--', alpha=0.5)
plt.legend()
plt.tight_layout()

# Exibir gráfico
plt.show()
```
